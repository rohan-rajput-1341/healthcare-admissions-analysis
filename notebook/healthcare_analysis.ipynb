# Healthcare Dataset Analysis
Cleaning, feature creation, exploration and charts before moving to SQL and Power BI.

## 1. Load the data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('healthcare_dataset.csv')   # on colab use '/content/healthcare_dataset.csv'
```

```python
df.head(10)
```

```python
df.info()
```

## 2. Cleaning

```python
# convert date columns from text to datetime
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
```

```python
# fix names (extra spaces and random capital letters)
df['Name'] = df['Name'].str.strip().str.title()
```

```python
# sort by admission date
df.sort_values('Date of Admission', inplace=True)
df.head(10)
```

```python
# quick checks
print(df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## 3. New columns

```python
# how many days the patient stayed
df['length_of_stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

# year and month of admission (needed for the line chart)
df['admission_year'] = df['Date of Admission'].dt.year
df['admission_month'] = df['Date of Admission'].dt.to_period('M')

df.head(10)
```

## 5. KPIs

```python
total_admissions = len(df)
total_revenue = df['Billing Amount'].sum()
avg_bill = df['Billing Amount'].mean()
avg_stay = df['length_of_stay'].mean()

print('Total admissions :', total_admissions)
print('Total revenue    :', round(total_revenue, 2))
print('Average bill     :', round(avg_bill, 2))
print('Average stay     :', round(avg_stay, 1), 'days')
```

## 6. Charts

### Test results

```python
test_counts = df['Test Results'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(test_counts, labels=test_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Test Results')
plt.show()
```

### Admission type

```python
adm_counts = df['Admission Type'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(adm_counts, labels=adm_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Admission Type')
plt.show()
```

### Blood type

```python
blood_counts = df['Blood Type'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(blood_counts.index, blood_counts.values)
plt.title('Patients by Blood Type')
plt.xlabel('Blood Type')
plt.ylabel('Number of Patients')
plt.show()
```

### Insurance provider

```python
ins_counts = df['Insurance Provider'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(ins_counts.index, ins_counts.values)
plt.title('Patients by Insurance Provider')
plt.xlabel('Insurance Provider')
plt.ylabel('Number of Patients')
plt.xticks(rotation=30)
plt.show()
```

### Medication

```python
med_counts = df['Medication'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(med_counts.index, med_counts.values)
plt.title('Medication Prescribed')
plt.xlabel('Medication')
plt.ylabel('Number of Patients')
plt.show()
```

### Admissions per month

```python
monthly = df.groupby('admission_month').size()
plt.figure(figsize=(12, 5))
plt.plot(monthly.index.astype(str), monthly.values)
plt.title('Admissions per Month')
plt.xlabel('Month')
plt.ylabel('Number of Admissions')
plt.xticks(range(0, len(monthly), 6), monthly.index.astype(str)[::6], rotation=45)
plt.show()
```

### Average bill by medical condition

```python
cond_bill = df.groupby('Medical Condition')['Billing Amount'].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 5))
plt.bar(cond_bill.index, cond_bill.values)
plt.title('Average Bill by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Billing Amount')
plt.xticks(rotation=30)
plt.show()
```
